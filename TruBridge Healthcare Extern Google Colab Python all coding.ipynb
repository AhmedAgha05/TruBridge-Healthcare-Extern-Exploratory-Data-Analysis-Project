```python
# 1. Keep just ONE value type - age-adjusted (per your research question)
df = df[df['DataValueTypeID'] == 'AgeAdjPrv']

# 2. Reshape long → wide: one row per county, each measure its own column
df_wide = df.pivot_table(
    index=['StateAbbr', 'LocationName', 'LocationID', 'TotalPopulation'],
    columns='MeasureId',
    values='Data_Value'
).reset_index()
df_wide = df_wide.rename(columns={
    'DIABETES': 'diabetes_prevalence',
    'ACCESS2': 'lack_of_insurance_rate',
    'HOUSINSECU': 'housing_insecurity_rate'
})
df_wide.head()
```
